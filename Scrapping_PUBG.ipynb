# Tahap 1 : Import Library Yang Dibutuhkan

```python
from google_play_scraper import app, reviews, Sort, reviews
import pandas as pd
pd.options.mode.chained_assignment = None
import numpy as np
import csv
```

```python
# Proses scrapping dataset terhadap aplikasi PUBG Mobile yang ada pada playstore dengan menggunakan google_play_scrapper
result, continuation_token = reviews(app_id = "com.tencent.ig",
                 lang = "en",
                 country = "us",
                 sort = Sort.NEWEST,
                 count = 15000)
```

```python
# Mengubah scrapping tersebut menjadi datafarame
df_scrapping = pd.DataFrame(result)
df_scrapping.head()
```

# Tahap 3: Melakukan Data Understanding

```python
# Melihat ukuran dataset
print("Jumlah ulasan:", df_scrapping.shape[0])
print("Jumlah kolom:", df_scrapping.shape[1])
```

```python
# Melihat jumlah nilai hilang
print("Jumlah nilai hilang adalah: ")
df_scrapping.isna().sum()
```

```python
# Melihat info umum dataset
print("Info umum dataset:")
df_scrapping.info()
```

# Tahap 4 : Simpan Dataset Tersebut Dengan Format CSV

```python
# Melakukan seleksi fitur yang relevan
df_scrapping = df_scrapping[['reviewId', 'content', 'score']]

# Menyimpan dataframe dalam format csv
with open('scrapping_pubg.csv', 'w', encoding='utf8', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(df_scrapping.columns)

    for row in df_scrapping.itertuples(index=False):
        writer.writerow(row)
```
